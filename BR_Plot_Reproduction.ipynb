# Plot reproduction

```python
import branching_ratios as br
import numpy as np
# constants
me = 0.511e-3 # GeV
mmu = 0.1057 # GeV
e = 0.303
c = 3e8 # m/s
mp = 0.9382720813
```

```python
# gauge coupling charges
gauge_couplings = {"A'":np.array([2/3,-1/3,-1/3,-1,-1,0,0,0]),
                   "B-L":np.array([1/3,1/3,1/3,-1,-1,-1,-1,-1]),
                   "B":np.array([1/3,1/3,1/3,-e**2/(4*np.pi)**2,-e**2/(4*np.pi)**2,0,0,0]),
                   "Chargephobic":np.array([-1/(3*np.sqrt(2)),np.sqrt(2)/3,np.sqrt(2)/3,0,0,-1/np.sqrt(2),-1/np.sqrt(2),-1/np.sqrt(2)])}
```

```python
# A' model
title = "A'"
alphaX = 1/137
x = gauge_couplings["A'"]
br.plot_branching_ratios(alphaX,x,title)
br.plot_decay_length(alphaX,x,title)
```

```python
# B-L model [x_u, x_d, x_s, x_e, x_mu, x_nu_e, x_nu_mu]
title = "B-L"
alphaX = 1/137
x = gauge_couplings["B-L"]
br.plot_branching_ratios(alphaX,x,title)
br.plot_decay_length(alphaX,x,title)
```

```python
# B model [x_u, x_d, x_s, x_e, x_mu, x_nu_e, x_nu_mu]
title = "B"
alphaX = 1/137
x = gauge_couplings["B"]
br.plot_branching_ratios(alphaX,x,title)
br.plot_decay_length(alphaX,x,title)
```

```python
# Chargephobic model [x_u, x_d, x_s, x_e, x_mu, x_nu_e, x_nu_mu]
title = "Chargephobic"
alphaX = 1/137
x = gauge_couplings["Chargephobic"]
br.plot_branching_ratios(alphaX,x,title)
br.plot_decay_length(alphaX,x,title)
```
